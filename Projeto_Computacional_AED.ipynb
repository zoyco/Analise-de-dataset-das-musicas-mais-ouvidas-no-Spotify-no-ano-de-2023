```r
install.packages("tidyverse")
install.packages("visdat")
install.packages("GGally")
library(tidyverse)
library(visdat)
library(GGally)
library(dplyr)
```

```r
df <- read.csv("https://github.com/HugoCarvalhoUFRJ/aed/raw/refs/heads/main/materiais-didaticos/spotify-2023-mod.csv", header = TRUE, sep = ',', dec = '.')
```

```r
df <- df %>%
  mutate(across(where(is.character), ~na_if(., "")))
```

```r
glimpse(df)
head(df)
summary(df)
```

```r
options(repr.plot.width = 10, repr.plot.height = 6)
ggplot(df,
       aes(x = danceability_.)) + geom_histogram(fill = 'green', color = 'black', bins = 73)
ggplot(df,
       aes(x = valence_.)) + geom_histogram(fill = 'green', color = 'black', bins = 93)
ggplot(df,
       aes(x = energy_.)) + geom_histogram(fill = 'green', color = 'black', bins = 88)
```

```r
cor(df$danceability_., df$energy_.)
ggplot(df,
       aes(x = danceability_., y = energy_.)) +
         geom_point()
```

```r
options(repr.plot.width = 25, repr.plot.height = 15)
df %>%
  select(danceability_. , energy_., valence_., bpm, acousticness_., instrumentalness_., liveness_., speechiness_., streams) %>%
  ggpairs()
```

```r
df %>%
  select(track_name,artist.s._name,danceability_.,valence_.,energy_.) %>%
  arrange(desc(energy_.))
```
